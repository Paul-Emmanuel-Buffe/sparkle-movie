# Sparkle movie
## Présentation:
Sparkle Movie est un projet d'analyse de données et d'exploration dédié à l'univers du cinéma. L'objectif est de collecter, nettoyer et analyser des jeux de données cinématographiques (films, notes, genres, box-office, équipes de production) afin d'en dégager des tendances clés et de poser les bases d'un futur moteur de recommandation ou d'aide à la décision.

Objectifs principaux
Exploration et nettoyage (EDA) : Structurer et fiabiliser les données brutes des films et des évaluations.

Analyse des tendances : Identifier les facteurs de succès (genres populaires, évolution des notes, impact des budgets et des sorties).

Modélisation / Recommandation : Préparer les variables nécessaires pour suggérer des films pertinents selon les préférences des spectateurs.
